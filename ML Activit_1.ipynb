```python
# Missing value summary

missing_data = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage": (df.isnull().mean() * 100).round(2)
})

missing_data = missing_data[
    missing_data["Missing Values"] > 0
]

missing_data.sort_values(
    by="Percentage",
    ascending=False
)
```

```python
# Preserve original dataset

original_df = df.copy()

# Create working copy

cleaned_df = df.copy()

print("Dataset copy created successfully!")
```

```python
# Handle missing numerical values

for col in cleaned_df.select_dtypes(
    include=["int64", "float64"]
).columns:

    cleaned_df[col] = cleaned_df[col].fillna(
        cleaned_df[col].median()
    )

print("Numerical missing values handled!")
```

```python
# Handle missing categorical values

for col in cleaned_df.select_dtypes(
    include=["object"]
).columns:

    if cleaned_df[col].isnull().any():

        cleaned_df[col] = cleaned_df[col].fillna(
            cleaned_df[col].mode()[0]
        )

print("Categorical missing values handled!")
```

```python
# Check remaining missing values

print("Total Missing Values:",
      cleaned_df.isnull().sum().sum())
```

```python
# Count duplicate rows

duplicates = cleaned_df.duplicated().sum()

print("Duplicate Rows:", duplicates)
```

```python
# Remove duplicate rows

cleaned_df = cleaned_df.drop_duplicates()

print("Dataset Shape:", cleaned_df.shape)
```

```python
# Display unique categorical values

for col in cleaned_df.select_dtypes(
    include=["object"]
).columns:

    print("\n", col)
    print(cleaned_df[col].unique())
```

```python
# Remove unnecessary spaces

for col in cleaned_df.select_dtypes(
    include=["object"]
).columns:

    cleaned_df[col] = cleaned_df[col].str.strip()

print("Categorical consistency checked!")
```

```python
# Create a copy for encoding

encoded_df = cleaned_df.copy()

# Select nominal categorical columns

columns_to_encode = [
    "Neighborhood",
    "HouseStyle",
    "BldgType",
    "SaleCondition"
]

# Apply one-hot encoding

encoded_df = pd.get_dummies(
    encoded_df,
    columns=columns_to_encode,
    dtype=int
)

print("Categorical encoding completed!")

encoded_df.head()
```

```python
from sklearn.preprocessing import StandardScaler

# Create a copy for scaling

scaled_df = encoded_df.copy()

# Select numerical features

features = [
    "LotArea",
    "GrLivArea",
    "TotalBsmtSF",
    "GarageArea"
]

# Apply standardisation

scaler = StandardScaler()

scaled_df[features] = scaler.fit_transform(
    scaled_df[features]
)

print("Numerical scaling completed!")

scaled_df[features].head()
```

```python
print("Final Dataset Shape:", scaled_df.shape)

print("\nRemaining Missing Values:")
print(scaled_df.isnull().sum().sum())

print("\nDuplicate Rows:")
print(scaled_df.duplicated().sum())

print("\nPrepared Dataset:")
display(scaled_df.head())
```

```python
# Save prepared dataset

scaled_df.to_csv(
    "house_price_prepared.csv",
    index=False
)

print("Prepared dataset saved successfully!")
```
